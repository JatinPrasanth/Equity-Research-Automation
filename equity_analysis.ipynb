```python
import yfinance as yf
import pandas as pd

ticker = yf.Ticker("AAPL")

income_statement = ticker.financials
balance_sheet = ticker.balance_sheet
cash_flow = ticker.cashflow

print("Income Statement")
display(income_statement)
```

```python
print("Balance Sheet")
display(balance_sheet)

print("Cash Flow Statement")
display(cash_flow)
```

```python
print("Income statement shape:", income_statement.shape)
print("Balance sheet shape:", balance_sheet.shape)
print("Cash flow shape:", cash_flow.shape)
```

```python
print("INCOME STATEMENT")
print(income_statement.index.tolist())

print("\nBALANCE SHEET")
print(balance_sheet.index.tolist())

print("\nCASH FLOW")
print(cash_flow.index.tolist())
```

```python
# Select the financial line items we need

financials = pd.DataFrame({
    "Revenue": income_statement.loc["Total Revenue"],
    "Operating Income": income_statement.loc["Operating Income"],
    "Net Income": income_statement.loc["Net Income"],
    "EPS": income_statement.loc["Diluted EPS"],
    "Total Assets": balance_sheet.loc["Total Assets"],
    "Total Debt": balance_sheet.loc["Total Debt"],
    "Equity": balance_sheet.loc["Stockholders Equity"],
    "Operating Cash Flow": cash_flow.loc["Operating Cash Flow"],
    "Capital Expenditure": cash_flow.loc["Capital Expenditure"],
    "Free Cash Flow": cash_flow.loc["Free Cash Flow"]
})

financials
```

```python
# Put the financial years into the index

financials = financials.T

financials.index.name = "Year"

financials
```

```python
financials = financials.T

financials.index = pd.to_datetime(financials.index).year
financials.index.name = "Year"

financials
```

```python
# Convert financial statement values to $ billions

money_columns = [
    "Revenue",
    "Operating Income",
    "Net Income",
    "Total Assets",
    "Total Debt",
    "Equity",
    "Operating Cash Flow",
    "Capital Expenditure",
    "Free Cash Flow"
]

financials[money_columns] = financials[money_columns] / 1_000_000_000

financials
```

```python
# Calculate financial performance metrics

financials["Revenue Growth"] = financials["Revenue"].pct_change() * 100

financials["Net Income Growth"] = financials["Net Income"].pct_change() * 100

financials["Operating Margin"] = (
    financials["Operating Income"] / financials["Revenue"] * 100
)

financials["Net Margin"] = (
    financials["Net Income"] / financials["Revenue"] * 100
)

financials["FCF Margin"] = (
    financials["Free Cash Flow"] / financials["Revenue"] * 100
)

financials["Debt to Equity"] = (
    financials["Total Debt"] / financials["Equity"]
)

financials
```

```python
# Return on Equity
financials["ROE"] = (
    financials["Net Income"] / financials["Equity"] * 100
)

# Return on Invested Capital
financials["ROIC"] = (
    financials["Operating Income"] / 
    (financials["Equity"] + financials["Total Debt"]) * 100
)

financials
```

```python
summary = financials[
    [
        "Revenue Growth",
        "Net Income Growth",
        "Operating Margin",
        "Net Margin",
        "FCF Margin",
        "Debt to Equity",
        "ROE",
        "ROIC"
    ]
].round(2)

summary
```

```python
financials = financials.sort_index()
financials.index.tolist()
```

```python
import matplotlib.pyplot as plt
import numpy as np

years = financials.index.astype(str)
x = np.arange(len(years))
width = 0.38

fig, ax = plt.subplots(figsize=(10, 6))

rev_bars = ax.bar(x - width/2, financials["Revenue"], width, label="Revenue", color="#1f3b63")
ni_bars = ax.bar(x + width/2, financials["Net Income"], width, label="Net Income", color="#6fa8dc")

ax.bar_label(rev_bars, fmt="%.0f", padding=3, fontsize=9)
ax.bar_label(ni_bars, fmt="%.0f", padding=3, fontsize=9)

ax.set_title("Apple Revenue vs Net Income (FY2022-FY2025)", fontsize=14, fontweight="bold")
ax.set_ylabel("USD billions")
ax.set_xticks(x)
ax.set_xticklabels(years)
ax.legend(frameon=False)
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.savefig("apple_revenue_vs_net_income.png", dpi=200)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(years, financials["Operating Margin"], marker="o", linewidth=2.2, color="#1f3b63", label="Operating Margin")
ax.plot(years, financials["Net Margin"], marker="o", linewidth=2.2, color="#6fa8dc", label="Net Margin")
ax.plot(years, financials["FCF Margin"], marker="o", linewidth=2.2, color="#e69138", label="FCF Margin")

for col, color in [("Operating Margin", "#1f3b63"), ("Net Margin", "#6fa8dc"), ("FCF Margin", "#e69138")]:
    for yr, val in zip(years, financials[col]):
        ax.annotate(f"{val:.1f}%", (yr, val), textcoords="offset points", xytext=(0, 8), ha="center", fontsize=9, color=color)

ax.set_title("Apple Margin Trends (FY2022-FY2025)", fontsize=14, fontweight="bold")
ax.set_ylabel("% of revenue")
ax.legend(frameon=False, loc="lower right")
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.savefig("apple_margin_trends.png", dpi=200)
plt.show()
```

```python
growth = financials[["Revenue Growth", "Net Income Growth"]].dropna()
gx = np.arange(len(growth))

fig, ax = plt.subplots(figsize=(10, 6))
b1 = ax.bar(gx - width/2, growth["Revenue Growth"], width, label="Revenue Growth", color="#1f3b63")
b2 = ax.bar(gx + width/2, growth["Net Income Growth"], width, label="Net Income Growth", color="#6fa8dc")

ax.bar_label(b1, fmt="%.1f%%", padding=3, fontsize=9)
ax.bar_label(b2, fmt="%.1f%%", padding=3, fontsize=9)
ax.axhline(0, color="black", linewidth=0.8)

ax.set_title("Apple YoY Growth (FY2023-FY2025)", fontsize=14, fontweight="bold")
ax.set_ylabel("% change")
ax.set_xticks(gx)
ax.set_xticklabels(growth.index.astype(str))
ax.legend(frameon=False, loc="upper left")
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.savefig("apple_growth.png", dpi=200)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

bars = ax.bar(years, financials["Free Cash Flow"], color="#1f3b63", width=0.55)
ax.bar_label(bars, fmt="%.1f", padding=3, fontsize=9)

ax.set_title("Apple Free Cash Flow (FY2022-FY2025)", fontsize=14, fontweight="bold")
ax.set_ylabel("USD billions")
ax.set_ylim(0, financials["Free Cash Flow"].max() * 1.15)
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.savefig("apple_fcf.png", dpi=200)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.plot(years, financials["ROE"], marker="o", linewidth=2.2, color="#1f3b63", label="ROE")
ax.plot(years, financials["ROIC"], marker="o", linewidth=2.2, color="#e69138", label="ROIC (simplified)")

for col, color in [("ROE", "#1f3b63"), ("ROIC", "#e69138")]:
    for yr, val in zip(years, financials[col]):
        ax.annotate(f"{val:.0f}%", (yr, val), textcoords="offset points", xytext=(0, 8), ha="center", fontsize=9, color=color)

ax.set_title("Apple Returns on Capital (FY2022-FY2025)", fontsize=14, fontweight="bold")
ax.set_ylabel("%")
ax.legend(frameon=False, loc="upper left")
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", alpha=0.25)

plt.tight_layout()
plt.savefig("apple_returns.png", dpi=200)
plt.show()
```
